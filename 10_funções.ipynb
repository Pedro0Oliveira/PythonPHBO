Para entregar números pares

```python
#função de retorno par
def epar(num):
    if num % 2 == 0:
        return True
    else:
        return False

valor = int(input("Digite um número inteiro para verificar se é primo: "))
if epar(valor):
    print(f"{valor} é um número par.")
else:
    print(f"{valor} não é um número par.")

#print(epar(4))
#print(epar(3))
```

Para entregar números impares

```python
#funcção para verificar se um número é primo
def primo(num):
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

valor = int(input("Digite um número inteiro para verificar se é primo: "))
if primo(valor):
    print(f"{valor} é um número primo.")
else:
    print(f"{valor} não é um número primo.")

#print(primo(4))
#print(primo(5))
```

Padrão de Fibonacci

```python
#função de Fibbonaci
def fibbonaci(n):
  seq = [1,1]
  for i in range(2,n):
    seq.append(seq[i-1]+seq[i-2])
  return seq
print(fibbonaci(12))
```
