Escopo de variáveis

```python
a=5
def muda_e_imprime():
  a=10
  print('a dentro da função: ', a) #a=10

print('a antes de mudar: ', a) #a=5
muda_e_imprime() #chamada da funcao, a=10
print('a depois de mudar: ', a) #a=5
```

```python
a=5
def muda_e_imprime():
  global a #declara que a é uma variável global
  a=10
  print('a dentro da função: ', a) #a=10

print('a antes de mudar: ', a) #a=5
muda_e_imprime() #chamada da funcao, a=10
print('a depois de mudar: ', a) #a=10
```

Função para embaralhar

```python
alunos = ['Daniel', 'Joao', 'Beatriz', 'Maria']
print(alunos)
random.shuffle(alunos)
print(alunos)
```

Função para embaralhar letras de uma palavra

```python
def embaralha(palavra):
  import random
  lista = list(palavra)
  print(lista)
  random.shuffle(lista)
  print(lista)
  return "".join(lista)

palavra = embaralha('programação')
print(palavra)
```

Funções Criptograficas

```python
def criptografa(msg):
  s = ''
  for c in msg:
    s += chr(ord(c)+1536)
    return s

def descriptografa(msg):
  s = ''
  for c in msg:
    s += chr(ord(c)-1536)
    return s

mensagem = "a turma de info é____"
msg_cripto = criptografa(mensagem)
print(msg_cripto)

print(descriptografa(msg_cripto))
```
